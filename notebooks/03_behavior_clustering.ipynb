# 03 - Behavior Clustering

Cluster trips into driving behavior types (aggressive, normal, cautious).

## Steps:
1. Load feature matrix
2. Perform clustering (KMeans or DBSCAN)
3. Analyze clusters
4. Save cluster labels


```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent / "src"))

import pandas as pd

from src.clustering import classify_driving_behavior, get_cluster_characteristics

# Load features
data_dir = Path("../data/processed")
features_df = pd.read_csv(data_dir / "trip_features_clustering.csv")

print(f"Loaded {len(features_df)} trips")
features_df.head()
```

## Perform Clustering

Classify trips into behavior types.


```python
# Perform KMeans clustering
feature_cols = [col for col in features_df.columns if col != "trajectory_id"]
clustered_df = classify_driving_behavior(
    features_df,
    method="kmeans",
    n_clusters=3,
    feature_cols=feature_cols
)

print("\nClustering results:")
print(clustered_df['behavior_type'].value_counts())

clustered_df.head()
```

## Cluster Characteristics

Analyze what makes each behavior type distinct.


```python
# Get cluster statistics
cluster_stats = get_cluster_characteristics(clustered_df)
print(cluster_stats)
```

## Save Results

Save cluster labels for visualization.


```python
# Save cluster assignments
output_file = data_dir / "trip_clusters.csv"
clustered_df[['trajectory_id', 'cluster', 'behavior_type']].to_csv(output_file, index=False)
print(f"Saved cluster assignments to {output_file}")
```
