# 01 - View Processed Data

This notebook displays the already-processed Geolife GPS trajectory data.

## What we'll do
1. Load the cleaned, processed data
2. Show summary statistics
3. Display a sample of the data
4. Visualize key features

**Note:** The data processing is done outside this notebook using `make process` or `make process-sample`.

```python
from pathlib import Path

import pandas as pd

# Set paths
DATA_DIR = Path("../data/processed")

# Load the cleaned, processed data
df = pd.read_parquet(DATA_DIR / "trajectories_cleaned.parquet")

print(f"✅ Loaded {len(df):,} GPS points")
print(f"📁 Unique users: {df['user_id'].nunique()}")
print(f"🛣️  Unique trajectories: {df['trajectory_id'].nunique()}")
print(f"📅 Date range: {df['timestamp'].min()} to {df['timestamp'].max()}")
```

## Summary Statistics

Key statistics on speed, acceleration, distance, and other features.


```python
# Summary statistics
print("📊 Speed statistics (km/h):")
print(df['speed_kmh'].describe())

print("\n📈 Acceleration statistics (m/s²):")
print(df['acceleration_ms2'].describe())

print("\n🗺️  GPS bounds:")
print(f"  Latitude:  [{df['lat'].min():.6f}, {df['lat'].max():.6f}]")
print(f"  Longitude: [{df['lon'].min():.6f}, {df['lon'].max():.6f}]")

print("\n📏 Distance & Duration per trip:")
print(df.groupby('trajectory_id')[['distance_km', 'duration_min']].mean().describe())
```
